### Setup

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import OllamaLLM
from langchain_core.output_parsers import StrOutputParser
#from langchain.chains import LLMChain (LLMChain was deprecated in LangChain v0.1.17 and removed from langchain.chains in newer versions.)
```

### Run the Ollama local server using docker

docker run -d --name ollama -p 11434:11434 -v ollama:/root/.ollama ollama/gemma3:1b

### Configure LLM and Promt Template. Create LLM chain

```python
llm = OllamaLLM(model="gemma3:1b", base_url="http://localhost:11434")
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("user", "What is {question} ?")
])
chain = prompt | llm | StrOutputParser()
response = chain.invoke({"question": "AI Agent"})
```

```python
## Print the response
print(response)
```
